# Specs and References

This module implements the EXPath Cryptographic Module specification. Several revisions of the spec exist, along with implementations in other XQuery engines. This chapter documents the sources and how they relate to the functions available in eXist-db.

## Specifications

### EXPath Cryptographic Module 1.0 (2010)

The original specification by Claudius Teodorescu.

- **URL:** [https://expath.org/spec/crypto](https://expath.org/spec/crypto)
- **Functions defined:** [`crypto:hash`]({docs}/functions/crypto/hash), [`crypto:hmac`]({docs}/functions/crypto/hmac), [`crypto:encrypt`]({docs}/functions/crypto/encrypt), [`crypto:decrypt`]({docs}/functions/crypto/decrypt), [`crypto:generate-signature`]({docs}/functions/crypto/generate-signature), [`crypto:validate-signature`]({docs}/functions/crypto/validate-signature)
- **Status:** All six functions are implemented in eXist-db

### EXPath Cryptographic Module Editor's Draft (2017)

An updated draft that consolidates function parameters into XPath 3.0 maps.

- **URL:** [https://expath.org/spec/crypto/editor](https://expath.org/spec/crypto/editor)
- **Key change:** `crypto:generate-signature($data, $parameters as map(*))` replaces multiple positional arities
- **Status:** The 2-param map-based `generate-signature` is supported alongside the original positional arities

### EXPath Crypto 2018 CG Final Report

A Community Group revision by Claudius Teodorescu adding provider introspection and algorithm naming updates.

- **URL:** [https://github.com/claudius108/expath-cg/blob/master/specs/crypto/crypto-new.html](https://github.com/claudius108/expath-cg/blob/master/specs/crypto/crypto-new.html)
- **Key additions:**
  - [`crypto:list-providers`]({docs}/functions/crypto/list-providers), [`crypto:list-services`]({docs}/functions/crypto/list-services), [`crypto:list-algorithms`]({docs}/functions/crypto/list-algorithms) — provider introspection functions (implemented but not yet enabled)
  - Java-style algorithm names: `DSAwithSHA1`, `RSAwithSHA1`
  - XML Canonicalization 1.1: `inclusive-1.1`, `inclusive-with-comments-1.1`

## Implementations

### BaseX

BaseX implements the crypto module with function signatures compatible with eXist-db's. Code written for BaseX's crypto functions should work on eXist-db without changes.

- **URL:** [https://docs.basex.org/main/Cryptographic_Functions](https://docs.basex.org/main/Cryptographic_Functions)
- **Functions:** [`crypto:hmac`]({docs}/functions/crypto/hmac), [`crypto:encrypt`]({docs}/functions/crypto/encrypt), [`crypto:decrypt`]({docs}/functions/crypto/decrypt), [`crypto:generate-signature`]({docs}/functions/crypto/generate-signature), [`crypto:validate-signature`]({docs}/functions/crypto/validate-signature)
- **Note:** BaseX does not implement [`crypto:hash`]({docs}/functions/crypto/hash) (use [`fn:hash`]({docs}/functions/fn/hash) instead). Algorithm names use lowercase (`sha256`, `md5`), which eXist-db also accepts.

### eXist-db expath-crypto-module 6.x (predecessor)

The previous implementation by Claudius Teodorescu, based on the external `crypto-java` library.

- **URL:** [https://github.com/eXist-db/expath-crypto-module](https://github.com/eXist-db/expath-crypto-module)
- **Status:** Superseded by v7.0 in eXist-db 7.0+. All v6 function arities are preserved for backward compatibility.
- **Breaking changes in v7:** [`crypto:encrypt`]({docs}/functions/crypto/encrypt) output changed from dash-separated bytes to `xs:base64Binary`; DSA_SHA1 signatures are forbidden by Java 21's secure validation.
