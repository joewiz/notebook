# What's Coming Next

eXist-db's JNode support is actively under development. This page describes features from the XQuery 4.0 specification that are planned but not yet implemented. The examples below show what will be possible — they will not run successfully today, but they illustrate the direction.

## `get()` as a Path Step

The XQuery 4.0 spec allows `get()` to be used as a step in a path expression. This enables positional lookups in arrays using path syntax:

```
(: NOT YET IMPLEMENTED — will work in a future release :)

let $matrix := [[1,2,3],[4,5,6],[7,8,9]]
return jtree($matrix)//get(2)

(: Would return: 2, 5, 8 — the second element of each nested array :)
```

Today, you can achieve the same result using explicit iteration:

````xquery
xquery version "4.0";

let $matrix := array { array {1,2,3}, array {4,5,6}, array {7,8,9} }
for $row in $matrix?*
return $row(2)
````

## JNode Kind Tests with Type Filters

The full spec allows filtering JNodes by the type of their value:

```
(: NOT YET IMPLEMENTED :)

let $tree := jtree(map { "name": "Ada", "age": 36, "active": true() })
return $tree//jnode(*, xs:integer)

(: Would return only the "age" node, since 36 is an integer :)
```

Today, you can filter by kind test and then check the value type:

````xquery
xquery version "4.0";

let $tree := fn:jtree(map { "name": "Ada", "age": 36, "active": true() })
for $child in $tree/child::*
where $child instance of number-node()
return fn:jkey($child) || " = " || fn:jvalue($child)
````

## The `gnode()` Kind Test

XQuery 4.0 introduces a **general node** concept (`gnode()`) that unifies XML nodes and JSON nodes under a single kind test. This would allow queries that seamlessly mix XML and JSON navigation:

```
(: NOT YET IMPLEMENTED :)

(: Find all "title" values whether the data is XML or JSON :)
let $xml := <library><book><title>Hamlet</title></book></library>
let $json := jtree(map { "book": map { "title": "Ulysses" } })
for $item in ($xml, $json)
return $item//gnode()[local-name(.) = "title" or jkey(.) = "title"]
```

Today, XML and JSON must be queried separately:

````xquery
xquery version "4.0";

let $xml-title := <library><book><title>Hamlet</title></book></library>//title/string()
let $json-title := fn:jvalue(fn:jtree(map { "book": map { "title": "Ulysses" } })/book/title)
return ("XML: " || $xml-title, "JSON: " || $json-title)
````

## Full Document Ordering Across JSON Trees

In the full spec, JNodes participate in document ordering alongside XML nodes. This enables operators like `<<` (precedes) and `>>` (follows) to work across JSON trees, and allows `union` (`|`) to merge XML and JSON result sets.

## Keyword Arguments with `collation()`

XQuery 4.0 introduces keyword arguments and a new [`fn:collation()`]({docs}/functions/fn/collation) function that returns a collation as a function item. This enables locale-aware sorting directly on JNode values:

```
(: NOT YET IMPLEMENTED :)

let $tree := jtree(map { "cities": array { "Zurich", "Aachen", "Orebro" } })
return $tree/cities/child::* => sort(collation(strength := "primary"))
```
