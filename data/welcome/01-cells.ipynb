# Cells and Serialization

The Notebook is an interactive environment for writing and running XQuery. Each document mixes prose with *cells* — fenced code blocks you can edit and execute in place.

## Declaring the Language Version

Cells default to XQuery 3.1. You can declare the version explicitly, which is useful both for documentation and to unlock version-specific features:

`````xquery
xquery version "3.1";

for $i in 1 to 5
return $i * $i
`````

To use XQuery 4.0 features — such as the pipeline operator `->` — just declare that the query is using XQuery 4.0:

`````xquery
xquery version "4.0";

(1, 2, 3, 4, 5) -> fn:sum(.)
`````

## Controlling the Output Format

By default, results use *adaptive* serialization: numbers look like numbers, strings like strings, XML like XML. You can request a specific format by adding options directly to the opening fence, in the form `{key=value}`.

In your markdown source, it looks like this:

````

`````xquery
<plays>{
    for $title in ("Hamlet", "Macbeth", "Othello")
    return <play>{$title}</play>
}</plays>
`````

````

Which renders as:

`````xquery
<plays>{
    for $title in ("Hamlet", "Macbeth", "Othello")
    return <play>{$title}</play>
}</plays>
`````

The available keys are:

- `method` — **adaptive** (default), **xml**, **json**, or **text**
- `indent` — **yes** or **no**
- `count` — maximum number of items to display (default: 1000)

You can also change the serialization method at runtime using the dropdown in the cell's toolbar — the fenced-block option just sets the default.
