# Sharing a Notebook

The Notebook lets you save the current state of your queries and share them with anyone via a short URL. The recipient does not need to be logged in to view and run a shared notebook — but you do need an account to create one.

## Who Can Share

Creating a share link requires your account to be a member of the **notebook-user** group (or the **notebook-manager** or **dba** groups). An administrator can add you to the group from the eXist-db user manager. If you see an error when you click **Share**, that is likely the cause.

## What Gets Saved

A share captures the *queries* in each cell as they stand at the moment you click **Share**: the text of every cell, its declared language version and serialization options, and whether **Use previous cells as context** is checked. What it does not capture is the *output* — results are never stored. When someone opens a shared link they see your queries ready to run, not a frozen snapshot of your results.

## Creating a Share Link

1. Edit and run your cells until the notebook is in the state you want to share.
2. Click the **Share** button in the notebook toolbar.
3. A share link is copied to your clipboard automatically. A brief confirmation message appears at the top of the page. Paste the link anywhere — a browser tab, an email, a chat message.

## Opening a Shared Link

Opening the link loads the saved queries into the same notebook layout. From there you (or the recipient) can run the cells, modify them, and create a new share of the revised version. Each share is independent — modifying a shared notebook never affects the original.

## A Note on Persistence

Shared states are stored as JSON documents in the database on the server that generated the link. They are identified by a UUID and are not attached to your account, so they cannot be edited or revoked after creation. Links are permanent for as long as the server continues to run the Notebook. If you want a revisable, named document you can return to, save it as a standalone notebook or a book chapter from the admin panel instead — shares are best thought of as snapshots you hand off, not living documents you maintain.
